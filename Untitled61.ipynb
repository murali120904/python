```python
#fibonacci series on tree recursions 0 1 1 2 3 5 8
def fib(n):
    if n<=1:
        return n
    return fib(n-1)+fib(n-2)
terms = int(input("enter the of terms to print:"))
print(("fibonacci series....."))
for i in range(terms):
    print(fib(i), end=' ')
```

```python
''' permuations of char in tree recursion'''
def permute(s,bucket=''):
    if not s:
        print(bucket)
        return
    for i in range(len(s)):
        ns=s[1:]+s[i+1:]
    permute(ns,bucket+s[0])
    permute(s[1:],bucket)
text=input("enter a name/word:")
print("possibilities of combinations.....")
permute(text)
```
